### Intro

I wanted to see what questions LLMs create for a date that go beyond the basic "What are your hobbies?", etc. I am interested to see what types of questions these are and if they can tell us a little more about what LLMs think we value when meeting someone new on a date.

I think that the questions an AI suggests might reveal something about what makes conversations actually work from patterns it's picked up from thousands of relationship advice articles, successful date stories, and conversational psychology. Are LLMs just regurgitating dating tips, or have they identified something fundamental about human connection?

I'll be using ChatGPT-5.1 for this experiment.

<mark style="background-color: lightblue; padding: 10px; display: block;">
PROMPT: I'm meeting someone new for coffee tomorrow. Instead of asking boring small talk questions, give me 5 interesting questions I could ask to really get to know them. Make them thoughtful but not too intense or weird.
</mark>

> Love that approach—good questions make coffee dates way better ☕️
>
> Here are **5 thoughtful, low-pressure questions** that invite real conversation without getting heavy or awkward:
>
> 1. **“What’s something you’ve been really into lately that most people wouldn’t guess about you?”**
>    → Opens the door to hobbies, quirks, or current interests.
>
> 2. **“What does a *perfectly ordinary* day look like for you?”**
>    → Tells you a lot about their values, pace of life, and what they enjoy day-to-day.
>
> 3. **“What’s the best piece of advice you’ve ever gotten—whether you follow it or not?”**
>    → Thoughtful, but still light and often leads to good stories.
>
> 4. **“What’s something you’re looking forward to in the next few months?”**
>    → Positive, future-oriented, and avoids pressure or big life questions.
>
> 5. **“What’s a small thing that instantly makes your day better?”**
>    → Sweet, revealing, and easy to build rapport from.
>
> If you want, I can tailor these to feel **more playful**, **more flirty**, or **more introspective** depending on the vibe you’re going for.


<mark style="background-color: lightblue; padding: 10px; display: block;">
PROMPT: why do you think these are good questions to ask?
</mark>

> These are good questions because they **invite depth without pressure**—they’re open-ended, personal, but still easy to answer. Here’s why they work well on a first coffee meet:
>
> > **They encourage stories, not one-word answers.**
> > Questions like “what are you into lately?” or “what’s a small thing that makes your day better?” naturally lead to anecdotes, which feel more human and engaging than facts.
>
> > **They reveal values indirectly.**
> > Asking about a “perfectly ordinary day” or what they’re looking forward to shows how they spend their time, what they prioritize, and what excites them—without asking something intense like “what do you value most in life?”
>
> > **They feel present-focused, not interrogative.**
> > These questions are about *now* and *near future*, which avoids résumé-style questions (job, major, five-year plan) that can feel stiff or evaluative.
>
> > **They create room for mutual sharing.**
> > Each question naturally invites you to answer too, which makes the conversation feel balanced and connected rather than like an interview.
>
> > **They build warmth and ease.**
> > Light curiosity (“small things,” “lately,” “ordinary days”) helps establish comfort and trust early—key ingredients for seeing if you actually enjoy each other’s company.


Looking at these five questions, I noticed some interesting patterns about what ChatGPT seems to think matters in getting to know someone:

1.  Present Over Past
Four out of five questions focus on right now or the near future. Only one asks about the past (the advice question). This is smart since asking about someone's past can feel like an interview or even invasive.

2. The Mundane Over the Monumental
"What does a perfectly ordinary day look like?" is smart because it's the opposite of what most people ask. We're trained to ask about highlights, but ordinary days reveal so much more.

3. Low Stakes, High Reward
None of these questions have a "wrong" answer. You can't fail at describing a small thing that makes your day better.

4. The Specificity Sweet Spot
"Something you've been into lately" is more specific than "hobbies" but broader than "what book are you reading?" It gives people options for how deep to go.


### Thoughts 

In my opinion, these are pretty good questions to get to know more about someone new. They really focus on the person and how they spend their time. 

The surprising thing? I think AI-generated questions might actually be better than what most of us come up with on our own, because we get nervous and default to safe, boring questions. LLMs don't get nervous. They just optimize for engagement. An AI that doesn't understand human connection can still help us connect better.
