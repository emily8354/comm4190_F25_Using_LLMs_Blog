### Chapter 8: AI as a Coach

Mollick breaks down how we develop expertise into three steps: gaining foundational knowledge, doing deliberate practice, and getting feedback. Traditionally, coaches and mentors guide people through this process, but there's what he calls an "apprenticeship gap". Entry-level work is getting automated, so new workers can't learn basic skills on the job anymore. At the same time, knowledge work is getting so complex that it is harder to transfer skills from experienced people to beginners. AI can potentially fill this gap by acting as a virtual coach that gives you immediate feedback, access to tons of information, and safe environments to practice through simulations.

What I found most interesting is how AI could democratize expertise. Right now, good coaching and mentorship are expensive and hard to access. If AI can provide personalized feedback and guidance to anyone, that could genuinely level the playing field for skill development. Mollick also points out that working with AI is creating a new kind of expertise itself. People who figure out how to use AI tools effectively in their field will have a real advantage. Over the summer, as an intern at a Big Tech company in the Bay Area, I constantly heard from leadership in the company that they were valuing "AI fluency" and expertise a lot in their employees and encouraging people to use AI to its full extent. However, I wonder whether AI coaching can really replace human mentors. There is something important about learning from someone who has been through it themselves, who can read your frustration or confusion and adjust their approach that I don't think AI can fully replicate.

### Chapter 9: AI as Our Future

The final chapter looks at four possible scenarios for AI's future. First is "As Good as It Gets," where AI has basically peaked and won't improve much more. Mollick thinks this is unlikely. Second is "Slow Growth," where AI keeps getting better but at a manageable pace that gives society time to adapt. Third is "Exponential Growth," where AI capabilities explode rapidly, bringing both huge opportunities and serious risks. Fourth is "The Machine God," where AI surpasses human intelligence entirely and we're dealing with something fundamentally more capable than us.

Mollick then walks through how AI will affect major parts of our lives regardless of which scenario plays out. Deepfakes and AI-generated content are already making it harder to trust what we see online. AI companions are getting sophisticated enough to address loneliness but might change human relationships in weird ways. The workplace is being transformed and some jobs will disappear, others will change completely, and we might need to rethink what "work" even means. On a bigger scale, AI could accelerate scientific discoveries and help solve global problems, but it's also concentrating massive power in the hands of a few tech companies. Mollick ends with a call to action: we need regulation, public education, and a focus on human-AI collaboration now, not later.

### Thoughts

The coaching chapter's focus on democratizing expertise is genuinely exciting. The idea that anyone could access personalized feedback and practice environments that were previously only available to people with money or connections could be transformative. In the future chapter, I appreciate that Mollick lays out multiple scenarios instead of pretending he knows what will happen. The exponential growth scenario feels most realistic to me based on what we've seen so far, which is both exciting and unsettling. I think the future of AI is mostly good, but improvements have happened so fast and I definitely think some regulation/thinking about what we are actually doing is necessary.
