> 🇲🇦 **1. Culture & Etiquette**
> Morocco is Muslim-majority; dress modestly in rural areas.
> Greetings like “Salam” or “Bonjour” are appreciated.
> Public affection is uncommon; Fridays are the holy day.
>
> 💬 **2. Languages**
> Arabic and Amazigh are official; French is widely spoken.
> English is common in tourist areas.
> Useful phrases: *Shukran* (thank you), *Salam* (hello), *La shukran* (no thank you).
>
> 💸 **3. Money**
> Currency: Moroccan Dirham (MAD).
> Cash is important in markets and taxis.
> ATMs are available; cards work in nicer places.
>
> 🧳 **4. Safety**
> Morocco is generally safe for tourists.
> Be cautious of petty scams and unsolicited “guides.”
> Keep belongings secure in crowded areas.
>
> 🕌 **5. Religion & Daily Life**
> Expect the call to prayer five times a day.
> During Ramadan, schedules shift and some restaurants close midday.
>
> 🧼 **6. Food & Water**
> Try tagine, couscous, pastilla, and mint tea.
> Drink bottled water, not tap.
> Choose street food with high turnover.
>
> 🚕 **7. Transportation**
> Petite taxis are cheap—ask for the meter.
> Grand taxis are shared for longer distances.
> Medinas like Marrakech are walk-heavy.
>
> 🛍️ **8. Souk & Bargaining Tips**
> Bargaining is expected; start at half the asking price.
> Be polite and patient when negotiating.
> Mint tea is hospitality, not a purchase obligation.
>
> ⚠️ **9. Photography Etiquette**
> Ask before photographing people.
> Some rural areas prefer no photos.
>
> ☀️ **10. Climate**
> Summers are very hot inland.
> Coastal cities are cooler.
> Desert nights get cold.

> Here are tips **specifically for Marrakech, Fes, and Chefchaouen**:
>
> 🇲🇦 **General**
> These three cities each feel completely different, so expect big contrasts.
> Keep small cash for taxis, markets, and tips.
> Medinas are maze-like—don’t rely only on Google Maps.
>
> 🔥 **Marrakech (Red City)**
> The medina is intense, loud, colorful, and busy—normal and part of the experience.
> Jemaa el-Fnaa square at night is chaotic but fun; watch for pickpockets.
> Visit Bahia Palace, Jardin Majorelle, and the Saadian Tombs.
> Taxis may try to overcharge—ask for the meter or agree on a price first.
> Stay near the medina walls if you want easier navigation.
>
> 🧿 **Fes (Fes el-Bali)**
> Fes has the *oldest, most traditional* medina in Morocco—expect narrow alleys.
> Hiring an official guide for half a day helps avoid getting lost or scammed.
> The Chouara Tannery is a must-see—the leather shops will give mint leaves to help with the smell.
> Shops are more traditional and less touristy than Marrakech.
> Avoid following anyone who says “this street is closed”—that’s a common line.
>
> 💙 **Chefchaouen (Blue City)**
> Much calmer than Marrakech and Fes—perfect for a relaxing end.
> Wander the blue streets early morning for the best photos and fewer crowds.
> Locals are friendly and more chill with tourists.
> The Spanish Mosque offers the best sunset spot.
> Food is slower-paced here—be patient and enjoy the vibe.
>
> 🚗 **Travel Between Cities**
> Train: Marrakech → Fes is easiest by ONCF train (fast and comfortable).
> Bus/Taxi: Fes → Chefchaouen usually requires a bus like CTM or a private transfer.
> Book transportation ahead during busy seasons.
>
> 🍽️ **Food Tips**
> Try tagine in Marrakech, traditional Fassi cuisine in Fes, and fresh local goat cheese in Chefchaouen.
> Drink bottled water everywhere.
> Tipping 5–10% is normal.
>
> 🧭 **Navigation & Safety**
> Medinas are safe but confusing—ignore unsolicited “guides.”
> If you get lost, ask a shopkeeper or a family, not random young guys.
> Dress modestly in Fes more than Marrakech; Chefchaouen is relaxed.

> Here are extra things most first-time visitors don’t realize until they’re already in Morocco:
>
> **1. Medinas can get overwhelming — take breaks**
> Marrakech and Fes are sensory-heavy; step into a rooftop café if you need calm.
>
> **2. Google Maps is unreliable inside old cities**
> Use it for general direction only; expect dead ends and unmarked alleys.
>
> **3. “Let me show you the way” is usually a hustle**
> Ask shopkeepers or older women if you’re lost.
>
> **4. Friday is the holy day**
> Some restaurants and shops may close midday.
>
> **5. Taxis rarely use meters unless you ask**
> Say “Compteur, s’il vous plaît?” or agree on a price beforehand.
>
> **6. Dress in layers**
> Days can be hot, nights can get chilly in Fes and Chefchaouen.
>
> **7. Moroccan time is relaxed**
> Food comes slowly; enjoy the pace.
>
> **8. Mint tea is hospitality**
> Accepting is polite, but you don’t have to buy anything afterward.
>
> **9. Bring small bills**
> Useful for tips, taxis, bathrooms, and markets.
>
> **10. Public affection is minimal**
> Holding hands is fine; anything more stands out.
>
> **11. Some bathrooms don’t flush toilet paper**
> Use the bin if that’s what locals do.
>
> **12. Chefchaouen photos need courtesy**
> Many “cute doorways” are real homes — ask or smile first.
