Hong Kong is surprisingly compact and has one of the world's best public transportation systems - everything is very accessible!

**Distances and travel times:**

Hong Kong Island to Kowloon is just across the harbour - the Star Ferry takes about 10 minutes, or the MTR (subway) is 2-5 minutes depending on which stations. Central to Tsim Sha Tsui is literally a 10-minute journey.

Within Hong Kong Island, Central to Causeway Bay is about 5-10 minutes by MTR. Central to Stanley (south side) takes 30-40 minutes by bus.

Within Kowloon, Tsim Sha Tsui to Mong Kok is about 10 minutes by MTR.

To the New Territories, you're looking at 30-45 minutes from central areas. For example, Central to Sai Kung takes about 45 minutes (MTR + minibus).

Lantau Island is about 30-35 minutes from Central via the Airport Express or Tung Chung line, then a cable car or bus to the Big Buddha. Other outlying islands require ferries - 30-60 minutes depending on the island.

**Getting around:**

The **MTR** is the backbone - clean, efficient, air-conditioned, with English signage. Trains run every few minutes. Get an Octopus card (rechargeable smart card) - it works on MTR, buses, trams, ferries, minibuses, and even at convenience stores.

**Buses** go everywhere the MTR doesn't, including scenic routes along the coast. Double-decker buses offer great views from the upper deck.

**Trams** ("ding dings") run along Hong Kong Island's north shore - they're slow but iconic and cheap, great for sightseeing.

**Minibuses** (green or red) fill gaps but can be confusing for visitors since routes aren't always clearly marked.

**Taxis** are relatively affordable - just note that Hong Kong Island, Kowloon, and New Territories have different colored cabs with different coverage areas.

**Star Ferry** is a must-do experience, not just transportation.

Google Maps works excellently in Hong Kong for public transport directions. You'll find getting around is really straightforward - the city is designed for easy navigation without a car!
