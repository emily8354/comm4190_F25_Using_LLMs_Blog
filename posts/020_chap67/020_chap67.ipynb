### Chapter 6: AI as a Coworker

Mollick starts the chapter with the fact that almost all of our jobs will overlap with the capabilities of AI. He tackles the uncomfortable reality that AI excels at some tasks we do while failing spectacularly at others. However, there are unclear boundaries between the two types of tasks, which Mollick calls the "Jagged Frontier." What struck me most was his framework for thinking about AI tasks: "Just Me Tasks" that you do yourself, "Delegated Tasks" where you assign work to AI but check it like an intern's output, and "Automated Tasks" where you let AI run unsupervised.

This resonates deeply with my own experience as I have caught myself both over-relying on AI for things it shouldn't handle and under-utilizing it for tasks where it could genuinely help. The research Mollick cites about consultants shows a real danger of "falling asleep at the wheel" and becoming so dependent on AI that our own skills go away. This is a legitimate concern that we need to actively manage.

What I appreciate about Mollick's approach is the pragmatism. He emphasizes that nearly all jobs will overlap with AI capabilities, but this doesn't mean wholesale replacement. It simply means the transformation of the tasks within those jobs. The key is figuring out where we add irreplaceable human value and where AI can genuinely augment our work.

### Chapter 7: AI as a Tutor

Chapter 7 addresses what Mollick calls the "Homework Apocalypse", the fact that AI can now complete most traditional assignments with ease. As someone currently in college, this chapter was both unsettling and exciting as I have seen firsthand how LLMs have changed my college experience with regards to homework.

Mollick invokes Benjamin Bloom's "two sigma problem", which is the finding that one-on-one tutoring dramatically improves student performance compared to traditional classroom settings. The possibility is that AI tutors could finally democratize this kind of personalized learning. Imagine every student having access to patient, adaptive tutoring that meets them exactly where they are. However, Mollick is not naive about the challenges. He emphasizes that traditional homework and essay assignments have become obsolete as measures of learning, forcing educators to fundamentally rethink assessment. Rather than fighting AI, he suggests focusing on higher-order thinking skills, emphasizing learning processes over final products, and even leveraging AI as a teaching tool itself.

What I found most interesting was Mollick's vision of flipped classrooms where AI handles content delivery, freeing classroom time for active learning, discussion, and relationship-building. This shifts teachers from "sage on the stage" to "guide on the side", a role that emphasizes the irreplaceable human elements of teaching: mentorship, emotional support, and helping students develop critical thinking. I am not sure how I feel about the flipped classroom method as I have taken courses at Penn with a flipped classroom and it really requires the students to learn by themselves or else they feel extremely lost during lectures and even more defeated. 

### Thoughts

Both chapters reinforce a central theme: AI isn't replacing humans, but it is forcing us to evolve. In the workplace, we need to become better at identifying where human judgment matters most. In education, we need to teach skills that AI can't replicate such as creativity, ethical reasoning, emotional intelligence, and the ability to work alongside these tools thoughtfully. The question isn't whether to engage with AI, but how to do so in ways that amplify rather than diminish our humanity.
